```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

everything will be lower case with underscore in between 

```python
df.columns=df.columns.str.lower()
```

```python
df.columns=df.columns.str.replace(' ','_')
```

```python
df=df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

create a column age_group

```python
# Create a column age_group
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df.head()
```

```python
df[['age','age_group']].head(10)
```

create another column purchase_frequency_days

```python
# create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
from sqlalchemy import create_engine
import urllib.parse  # 1. Import the URL safety utility

# Step 1: Connect to PostgreSQL
username = "postgres"        
password = "Hi@221004" 
host = "localhost"           
port = "5432"               
database = "customer_behavior"  

# 2. Encode the password so the '@' sign does not break the connection string
safe_password = urllib.parse.quote_plus(password)

# 3. Use 'safe_password' inside your connection string instead of 'password'
engine = create_engine(f"postgresql+psycopg2://{username}:{safe_password}@{host}:{port}/{database}")

# Step 2: Load DataFrame into PostgreSQL
table_name = "customer"     
df.to_sql(table_name, engine, if_exists="replace", index=False)

print(f"Data successfully loaded into table '{table_name}' in database '{database}'.")
```
